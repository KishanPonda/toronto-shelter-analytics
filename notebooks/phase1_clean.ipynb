## Imports

```python
# Import pandas for data manipulation and os for file path handling

import pandas as pd
import os
```

## File Paths

```python
# Define each file name and its date format
# 2021 and 2022 use short year format (YY-MM-DD), rest use full ISO format

files = {
    "2021": ("2021_Data.csv", "%y-%m-%d"),
    "2022": ("2022_Data.csv", "%y-%m-%d"),
    "2023": ("2023_Data.csv", "mixed"),
    "2024": ("2024_Data.csv", "mixed"),
    "2025": ("2025_Data.csv", "mixed"),
    "2026": ("2026_Data_30th June.csv", "mixed"),
}
```

## Displaying First 3 Rows for each year before Cleaning

```python
# Quick sanity check — display first 3 rows of each file
# to confirm files loaded correctly before full processing

for year, (path, fmt) in files.items():
    df = pd.read_csv(path, nrows=3)
    print(f"\n=== {year} ===")
    display(df.head(3))
```

## Loading and Parsing Each file

```python
# Load all 6 files completely, parse dates with correct format per year
# Normalize removes the time component and keeps date only

dfs = []
for year, (path, fmt) in files.items():
    df = pd.read_csv(path)
    df['OCCUPANCY_DATE'] = pd.to_datetime(df['OCCUPANCY_DATE'], format=fmt)
    df['OCCUPANCY_DATE'] = df['OCCUPANCY_DATE'].dt.normalize()
    print(f"{year}: {len(df):,} rows")
    dfs.append(df)
```

## Concatenate and deduplicate

```python
# Stack all 6 dataframes into one and remove any exact duplicate rows

combined = pd.concat(dfs, ignore_index=True)
before = len(combined)
combined = combined.drop_duplicates()
print(f"Before dedup : {before:,}")
print(f"After dedup  : {len(combined):,}")
print(f"Removed      : {before - len(combined):,}")
```

## Validate date range

```python
# Confirm all dates fall within the expected 2021-2026 window
# If any date is outside this range the script will raise an error

min_date = combined['OCCUPANCY_DATE'].min()
max_date = combined['OCCUPANCY_DATE'].max()
print(f"Date range: {min_date.date()} → {max_date.date()}")
assert min_date.year == 2021
assert max_date.year == 2026
print("Date range check: PASSED")
```

## Adding Derived Fields

```python
# CAPACITY_GAP: how many funded beds are offline each night per program
# AT_RISK_FLAG: True if program is overcrowded (>90%) or underutilized (<40%)

combined['CAPACITY_GAP'] = (
    combined['CAPACITY_FUNDING_BED'] - combined['CAPACITY_ACTUAL_BED'])

combined['AT_RISK_FLAG'] = (
    (combined['OCCUPANCY_RATE_BEDS'] > 90) |
    (combined['OCCUPANCY_RATE_BEDS'] < 40))

print(f"CAPACITY_GAP avg : {combined['CAPACITY_GAP'].mean():.2f} beds")
print(f"AT_RISK_FLAG True: {combined['AT_RISK_FLAG'].sum():,} rows")
```

## Exporting the Clean CSV

```python
# Save the final cleaned dataset to the current working directory
# This file will be used in Phase 2 to load into MySQL

combined.to_csv("shelter_clean.csv", index=False)
print(f"Saved: shelter_clean.csv")
print(f"Rows : {len(combined):,}")
print(f"Cols : {len(combined.columns)}")
print("Phase 1 complete.")
```
